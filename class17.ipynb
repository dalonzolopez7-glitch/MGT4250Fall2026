```python
# Create a Streamlit application that contains 
# three visualizations from “airbnb.xlsx”. 
# You need to search online for some options

import pandas as pd
import plotly.express as px
df = pd.read_excel("airbnb.xlsx")
```

```python
df.columns
```

```python
# Figure 1: A bar plot showing the number of hosts by neighborhood. 
c = df.groupby("Neighbourhood ")["Host Id"].nunique().\
    reset_index(name="Number of hosts")
fig1 = px.bar(
    c,
    x="Neighbourhood ",
    y="Number of hosts"
)
fig1.show()
```

```python
# Figure 2: A box plot showing the distribution of prices by Room Type. 
# Color by Room Type.
# Change the scale of Y-axis to a Log scale
fig2 = px.box(
    df, 
    x="Room Type",
    y="Price",
    log_y=True,
    color="Room Type"
)
fig2.show()
```

```python
# Figure 3: A scatter plot of which X-axis is Number Of Reviews 
# and Y-axis is Price. Color by Room Type.
# Change the scale of Y-axis to a Log scale

fig3 = px.scatter(
    df,
    x="Number Of Reviews",
    y="Price",
    color="Room Type",
    log_y=True
)
fig3.show()
```
